```python
import numpy as np
import math
data = np.genfromtxt('/content/iris_1D.csv',
                     delimiter=',',
                     dtype= None)

data
```

```python
data = data[1:]  #Skip dòng đầu
x_data = data[:,0]
y_data = data[:,1]
```

```python
def sigmoid_function(z):
  """ Tính giá trị z qua hàm logistic"""
  return 1 / (1 + math.exp(-z))

def loss_function(y_hat, y):
  loss = - y*math.log(y_hat) - (1 - y)*math.log(1-y_hat)
  return loss

def predict(x,w,b):
  """Chúng ta sẽ tính giá trị dự đoán qua 2 bước, bước 1 tính z qua mối quan hệ tuyến tính với w và b,
  bước 2 chúng ta sẽ tính giá trị y_hat qua hàm logistic"""
  z = w*x + b
  y_hat = sigmoid_function(z)
  return y_hat

def compute_gradient(y, y_hat,x):
  dw = x*(y_hat - y)
  db = y_hat - y
  return dw, db

def update_weights(lr,w,b,y,y_hat):
  dw,db = compute_gradient(y,y_hat,x)
  w = w - lr*dw
  b = b - lr*b
  return w,b
```

```python
# some parameters
lr = 0.01
num_epochs = 1

w = -0.1
b = 0.1

for _ in range(num_epochs):
    for i in range(6):
        # get data
        x = x_data[i]
        y = y_data[i]

        # Correctly assign the float conversion back to variables
        x = float(x)
        y = float(y)
        print('x, y:', x, y)

        # predict
        y_hat = predict(x, w, b)
        print('y_hat:', y_hat)

        # compute loss
        loss = loss_function(y_hat, y)
        print('loss:', loss)

        # compute gradient
        db, dw = compute_gradient(y, y_hat,x)
        print('db, dw:', db, dw)

        # update
        b, w = update_weights(lr,w,b,y,y_hat)
        print('b, w:', b, w)
        #print('----------')
        break
```
