```python
# sphere function
#khởi tạo hàm f(x) = x ^2
def func(x):
    return x**2
#khởi tạo hàm d(f(x))_dx
def derivative(x):
    return 2*x
```

```python
import numpy as np #import thư viện numpy
import matplotlib.pyplot as plt  #import thư viện matplotlib

### Ý tưởng ở đây của chúng ta là sẽ khởi tạo 1 sequence trong khoảng từ -100, 100 bước nhảy 0.1 để quán sát hỉnh dạng của hàm f(x)

x = np.arange(-100, 100, 0.1)
y = func(x)

plt.plot(x, y)
plt.show()
```

```python
import random

num_steps = 10
step = 1
lr = 0.1

# 1. set x randomly
x = random.randint(-100, 100)
print(x)

for _ in range(num_steps):
    # compute derivative
    dx = derivative(x)

    # update
    x = x - lr*dx
    print(x, func(x))
```
